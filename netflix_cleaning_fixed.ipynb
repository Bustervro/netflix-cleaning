```python
import pandas as pd

# PART 1: Missing Data
netflix_titles = pd.read_csv("netflix_titles.csv")

print("Original shape:", netflix_titles.shape)

print("\nMissing values:")
print(netflix_titles.isnull().sum())

missing_percent = netflix_titles.isnull().mean() * 100

print("\nMissing percentages:")
print(missing_percent.round(2))

print("\nColumns with most missing data:")
print(missing_percent.sort_values(ascending=False).head(3))


# PART 2: Cleaning Data

# Strategy A: Fill missing directors
netflix_filled = netflix_titles.copy()

netflix_filled["director"] = netflix_filled["director"].fillna("Unknown")

print("\nMissing directors after filling:")
print(netflix_filled["director"].isnull().sum())

# Strategy B: Drop missing rows
netflix_dropped = netflix_titles.dropna(
    subset=["director", "cast"]
)

print("\nShape after dropping:")
print(netflix_dropped.shape)

rows_lost = len(netflix_titles) - len(netflix_dropped)
print("Rows lost:", rows_lost)


# PART 3: Date Conversion

print("\nDate type before:")
print(netflix_filled["date_added"].dtype)

netflix_filled["date_added"] = pd.to_datetime(
    netflix_filled["date_added"].str.strip(),
    errors="coerce"
)

print("\nData information after conversion:")
netflix_filled.info()

netflix_filled["day_added"] = netflix_filled["date_added"].dt.day

print("\nFirst five rows:")
display(netflix_filled.head())
```
